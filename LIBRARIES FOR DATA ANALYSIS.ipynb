```python
### merging and joining Dataframe
dt1=pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
dt2=pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
dt1
```

```python
dt2
```

```python
## Merge Dataframe on the 'Key Columns'
pd.merge(dt1,dt2,on='Key',how='inner')  #Only matching rows in both
```

```python
pd.merge(dt1,dt2,on='Key',how='right') #All rows from dt2, matching from dt1
```

```python
pd.merge(dt1,dt2,on='Key',how="left") #all of dt1, matching from dt2
```

```python
pd.merge(dt1,dt2,on='Key',how="outer")  #All rows from both, NaN where no match
```

### MATPLOTLIB

```python
import matplotlib.pyplot as plt
```

```python
# create first chart
x=[1,2,3,4,5]
y=["A","B",'C','D','F']

plt.plot(x,y)    #line chart
plt.show()      #displaying chart
```

```python
# customize chart
plt.figure(figsize=(9,4))
plt.plot(x,y,color='y',linestyle="--",marker='o',linewidth=2,markersize=14)
plt.title("Basic Line Chart")
plt.xlabel("Number of Students")
plt.ylabel("grades")
plt.grid(False)   #remove background lines
```

```python
# Advanced line chart - Multiple lines and legends
x=[1,2,3,4,5]
y1=["A","B",'C','D','F']
y2=['b+','c-','c+','A-','D+']

plt.plot(x,y1,label='fall 24 grades')
plt.plot(x,y2,label='fall 25 grades')

plt.title("Student grades")
plt.xlabel("Months")
plt.ylabel("Grades")

plt.legend()
plt.grid(False)
```

```python
# Bar chart
x=[1,2,3,4,5]
y=["A","B",'C','D','F']

plt.bar(x,y)
plt.title("Student Grades")
plt.xlabel("Months")
plt.ylabel("Grades")

plt.show()
```

```python
#histogram
#used for distribution analysis
import random

data=[random.randint(1,30) for x in range(20)]
print(data)
plt.hist(data,bins=6)       #bins=divide data into 6 parts
plt.title("Histogram")
plt.grid(False)
```

```python
#pie chart used to show relationships/data
region=['kpk','punjab','sindh','balochistan']
population=[23,45,12,24]
colors=['grey','pink','orange','green']
explode=(0,0,0,0.2) ##move out the 4th slice
plt.pie(population,explode=explode,labels=region,colors=colors,autopct='%1.1f%%',startangle=90)
plt.title("Population Histogram")
plt.show()
```

```python
#Scatter plot
#used to find relationship between variables
y1=[4,7,3,9,8]
y2=[3,5,8,1,9]

plt.scatter(y1,y2)
plt.title("Scatter Plot")
plt.grid(False)
plt.show()
```

```python
#subplots
#useed to show multiple charts in oe figure

#bar chart
categories = ['Mon','Tue','Wed','Thu','Fri']
sales = [10, 20, 55, 35, 45]

#scatter plot
y1 = [10, 20, 25, 35, 45]
y2 = [20, 30, 35, 45, 55]

plt.figure(figsize=(10,4))

plt.subplot(1,2,2)     #row,column,position
plt.bar(categories,sales)
plt.xlabel("Categories")
plt.ylabel("Sales")
plt.title("Bar Chart")
plt.grid(False)

plt.subplot(1,2,1)
plt.scatter(y1,y2)
plt.xlabel("user1")
plt.ylabel("user2")
plt.title("Scatter Plot")
plt.grid(False)
plt.show()
```
